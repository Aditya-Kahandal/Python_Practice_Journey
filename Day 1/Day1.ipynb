```python
def ReverseNumber(n):
  reverse = 0
  while n>0:
    digit = n%10
    reverse = reverse*10 + digit
    n = n//10
  print(reverse)

number = int(input("Enter a number: "))
ReverseNumber(number)
```

```python
def isPalindrome(n):
  rev = 0
  number = n
  while n > 0:
    digit = n%10
    rev = rev*10+ digit
    n = n//10
  if number == rev:
    print("The number is a Palindrome")
  else:
    print("The number is not Palindrome")

number = int(input("Enter a number: "))
isPalindrome(number)
```

```python
def countDigits(n):
  count = 0
  while n > 0:
    n = n//10
    count = count + 1
  return count

number = int(input("Enter a number: "))
print(countDigits(number))
```

```python
def Factorial(n):
  fact = 1
  if n == 0:
    return 1
  elif (n<0):
    return -1
  else:
    while n>0:
      fact = fact * n
      n = n-1
    return fact

number = int(input("Enter a number: "))
Factorial(number)
```

```python
# Bonus Question

def Star_Pattern(n):
  for i in range ( 0, n):
    for j in range (0, i+1):
      print("*" , end = " ")
    print ("\n")



number = int(input("Enter a number: "))
Star_Pattern(number)
```

```python
def isPrime(n):
  if n==1 or n ==2 :
    return True

  for i in range(2,n):
    if n%i == 0:
      return False
    else:
      continue
  return True

number = int(input("Enter a number: "))
if isPrime(number): print("Prime")
else: print("Not Prime")
```
