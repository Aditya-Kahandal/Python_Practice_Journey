```python
#Odd Even Checker

def OddEven(number):
    if number % 2 == 0:
        return "Number is Even"
    else:
        return "Number is Odd"

print(OddEven(int(input("Enter a Number:"))))
```

```python
# Number Guessing Game

'''
steps:
1) decide a secret number manually
2) initialize a counter = 1 and guess
3) create a function which compares the guess to the secret number and continues till the  guess is right
4) return counts of guesses and the result

'''
def guessTheNumber(key):
    guess = int(input("Enter your guess:"))
    counter = 1
    while guess != key:
        counter +=1
        new_guess = int(input("Wrong guess! Try again: "))
        guess = new_guess
    if counter==1:
      print("\nBrilliant, You guessed it in First Try")
    else:
      print("\nYou got it !!")
      print ("You took", counter,"guesses")

print("Welcome to the Ussy_Plays_Guessy:\nTry to Guess my Secret Number")
secret_num = 13

guessTheNumber(secret_num)
```

```python
# Frequency Counter

"""
Approach:
1) Use of dictionary
2)traverse through the string,
3)if the character is present in the string, add the count
4) else assign value of the character as 1 and add to the list
"""

def freq_Count(word):
    res = {}
    for char in word:
        if char in res:
            res[char] += 1
        else:
            res[char] = 1

    return res

input_word = input("Enter the word:")
print(freq_Count(input_word))
```

```python
#Student Marks Analyzer:

"""
Approach:
1) take the array as input to the functions for average, highest, lowest and passcount
2) implement average of marks
3) implement finding highest marks
4) implement finding the lowest marks
5) implement pass count of subjects
"""

def average(arr):
    length = len(arr)
    sum = 0
    for i in range(len(arr)):
        sum += arr[i]
    return sum/length

def highest(arr):
    length = len(arr)
    if length == 0:
        return -1
    highest = arr[0]
    for i in range(1,len(arr)):
        if arr[i]> highest:
            highest = arr[i]
    return highest

def lowest(arr):
    length = len(arr)
    if length == 0:
        return -1
    lowest = arr[0]
    for i in range(1,len(arr)):
        if arr[i]< lowest:
            lowest = arr[i]
    return lowest

def pass_count(arr):
    count = 0
    for marks in arr:
        if marks > 40:
            count+= 1

    return count
list_of_marks= [78,65,90,88,39]

print("Average of the subjects is:", average(list_of_marks))
print("Highest marks from subjects is:", highest(list_of_marks))
print("Lowest marks from subjects is:", lowest(list_of_marks))
print("You have passed in", pass_count(list_of_marks),"subjects")
```
